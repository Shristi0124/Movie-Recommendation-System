Importing the Dependencies

```python
import numpy as np
import pandas as pd
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

Data Collectiion and pre-Preocessing

```python
#loading the data from the csv file to apandas dataframe
movies_data = pd.read_csv('/content/movies.csv')
```

```python
#printing the first 5 rows of the dataframe
movies_data.head()
```

```python
#number of rows and coloue in dataset
movies_data.shape
```

```python
#selecting in relevant fetuture for recammendation
selected_features = ['genres','keywords','tagline','cast','director']
print(selected_features)
```

```python
# replacing the null values with null string

for feature in selected_features:
  movies_data[feature] =movies_data[feature].fillna('')
```

```python
# combining all the selected features

combined_features = movies_data['genres']+ ' '+movies_data['keywords']+ '' +movies_data['tagline']+'' +movies_data['cast']+ '' +movies_data['director']
```

```python
print(combined_features)
```

```python
# conversting the text data to feature vectors
vectorizer = TfidfVectorizer()
```

```python
feature_veators = vectorizer.fit_transform(combined_features)
```

```python
print(feature_veators)
```

Cosine Similarity

```python
# getting the similarity scores using cosine similarity

similarity = cosine_similarity(feature_veators)
```

```python
print(feature_veators)
```

```python
print(similarity.shape)
```

```python
# getting the movie name for user
movie_name = input('Enter the favourite movies name : ')
```

```python
#cerating  the list of movies all the name givien in the dataset

list_of_all_titles = movies_data['title'].tolist()
print(list_of_all_titles)
```

```python
#finding the close match for the movies name given by the user

find_close_match = difflib.get_close_matches(movie_name,list_of_all_titles)
print(find_close_match)
```

```python
close_match = find_close_match[0]
print(close_match)
```

```python
# finding the index of the movies with title

index_of_the_movie = movies_data[movies_data.title == close_match]['index'].values[0]
print(index_of_the_movie)
```

```python
# getting a list of similar movies

similarity_score = list(enumerate(similarity[index_of_the_movie]))
print(similarity_score)
```

```python
len(similarity_score)
```

```python
# sorting the  movies based on their similarity score

sorted_similar_movies = sorted(similarity_score,key = lambda x:x[1],reverse = True)
print(sorted_similar_movies)
```

```python
# print the name of similar movies based on the index
print ('Moves Suggest for you :')
i=1
for movie in sorted_similar_movies:
  index = movie[0]
  title_from_index = movies_data[movies_data.index==index]['title'].values[0]
  if (i<30):
    print(i,'.',title_from_index)
    i+=1
```

Movie Rcommendation System

```python
import gradio as gr
import pandas as pd
import difflib
import pickle  # or joblib
def recommend_movies(movie_name):
    try:
        list_of_all_titles = movies_data['title'].tolist()
        find_close_match = difflib.get_close_matches(movie_name, list_of_all_titles)

        if not find_close_match:
            return "❌ No similar movie found. Try a different name."

        close_match = find_close_match[0]
        index_of_the_movie = movies_data[movies_data.title == close_match]['index'].values[0]

        similarity_score = list(enumerate(similarity[index_of_the_movie]))
        sorted_similar_movies = sorted(similarity_score, key=lambda x: x[1], reverse=True)

        recommended_movies = []
        for i, movie in enumerate(sorted_similar_movies[1:31], start=1):
            index = movie[0]
            title_from_index = movies_data[movies_data.index == index]['title'].values[0]
            recommended_movies.append(f"{i}. {title_from_index}")

        return "\n".join(recommended_movies)

    except Exception as e:
        return f"⚠️ Error: {str(e)}"
```

```python
gr.Interface(
    fn=recommend_movies,
    inputs=gr.Textbox(label="Enter your favourite movie"),
    outputs=gr.Textbox(label="Top 30 Recommended Movies"),
    title="🎬 Movie Recommender",
    description="Type a movie name to get similar movie recommendations."
).launch()
```
